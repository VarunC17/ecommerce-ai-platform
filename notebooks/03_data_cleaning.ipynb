# 03 - Data Cleaning

This notebook cleans the Olist e-commerce datasets.

Cleaning tasks include:

- Checking dataset structure
- Checking missing values
- Removing duplicate records
- Handling missing values
- Converting data types
- Validating date columns
- Creating useful derived columns
- Saving cleaned datasets to the processed data folder

Input:
data/raw/

Output:
data/processed/

```python
import os
import warnings

warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
PROJECT_ROOT = r"C:\Users\asus\Documents\ECOMMERCE-AI-PLATFORM"

DATA_DIR = os.path.join(PROJECT_ROOT, "data")

RAW_DATA_DIR = os.path.join(DATA_DIR, "raw")

PROCESSED_DATA_DIR = os.path.join(DATA_DIR, "processed")

print("Project paths configured successfully!")
print(f"Raw data directory: {RAW_DATA_DIR}")
print(f"Processed data directory: {PROCESSED_DATA_DIR}")
```

```python
os.makedirs(PROCESSED_DATA_DIR, exist_ok=True)

print("Processed data directory is ready!")
```

```python
customers = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_customers_dataset.csv")
)

geolocation = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_geolocation_dataset.csv")
)

order_items = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_order_items_dataset.csv")
)

order_payments = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_order_payments_dataset.csv")
)

order_reviews = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_order_reviews_dataset.csv")
)

orders = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_orders_dataset.csv")
)

products = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_products_dataset.csv")
)

sellers = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "olist_sellers_dataset.csv")
)

category_translation = pd.read_csv(
    os.path.join(RAW_DATA_DIR, "product_category_name_translation.csv")
)

print("All datasets loaded successfully!")
```

```python
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}

print(f"Total datasets loaded: {len(datasets)}")
```

```python
print("Dataset Shapes")
print("-" * 50)

for name, dataframe in datasets.items():
    print(f"{name:<25} {dataframe.shape}")
```

```python
for name, dataframe in datasets.items():

    print("=" * 80)
    print(f"DATASET: {name}")
    print("=" * 80)

    print("\nColumns:")
    print(dataframe.columns.tolist())

    print("\nData Types:")
    print(dataframe.dtypes)

    print()
```

```python
print("Missing Values Before Cleaning")
print("=" * 80)

for name, dataframe in datasets.items():

    missing_values = dataframe.isnull().sum()

    missing_values = missing_values[missing_values > 0]

    print(f"\n{name}")

    if len(missing_values) == 0:
        print("No missing values found.")

    else:
        print(missing_values)
```

```python
print("Duplicate Rows Before Cleaning")
print("=" * 80)

for name, dataframe in datasets.items():

    duplicate_count = dataframe.duplicated().sum()

    print(f"{name:<25} {duplicate_count}")
```

```python
for name in datasets:

    datasets[name] = datasets[name].drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed successfully!")
```

```python
customers = datasets["customers"]

geolocation = datasets["geolocation"]

order_items = datasets["order_items"]

order_payments = datasets["order_payments"]

order_reviews = datasets["order_reviews"]

orders = datasets["orders"]

products = datasets["products"]

sellers = datasets["sellers"]

category_translation = datasets["category_translation"]

print("Dataset variables updated successfully!")
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:

    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

print("Order date columns converted successfully!")
```

```python
print(orders[date_columns].dtypes)
```

```python
order_reviews["review_creation_date"] = pd.to_datetime(
    order_reviews["review_creation_date"],
    errors="coerce"
)

order_reviews["review_answer_timestamp"] = pd.to_datetime(
    order_reviews["review_answer_timestamp"],
    errors="coerce"
)

print("Review date columns converted successfully!")
```

```python
orders["order_approved_at"] = orders["order_approved_at"].fillna(
    orders["order_purchase_timestamp"]
)

orders["order_delivered_carrier_date"] = (
    orders["order_delivered_carrier_date"]
    .fillna(orders["order_approved_at"])
)

orders["order_delivered_customer_date"] = (
    orders["order_delivered_customer_date"]
    .fillna(orders["order_estimated_delivery_date"])
)

print("Missing order date values handled successfully!")
```

```python
product_numeric_columns = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for column in product_numeric_columns:

    products[column] = products[column].fillna(
        products[column].median()
    )

products["product_category_name"] = (
    products["product_category_name"]
    .fillna("unknown")
)

print("Missing product values handled successfully!")
```

```python
order_reviews["review_comment_title"] = (
    order_reviews["review_comment_title"]
    .fillna("No Title")
)

order_reviews["review_comment_message"] = (
    order_reviews["review_comment_message"]
    .fillna("No Comment")
)

print("Missing review text values handled successfully!")
```

```python
products["product_category_name"] = (
    products["product_category_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

category_translation["product_category_name"] = (
    category_translation["product_category_name"]
    .astype(str)
    .str.strip()
    .str.lower()
)

category_translation["product_category_name_english"] = (
    category_translation["product_category_name_english"]
    .astype(str)
    .str.strip()
    .str.lower()
)

print("Product category names standardized successfully!")
```

```python
customers["customer_city"] = (
    customers["customer_city"]
    .astype(str)
    .str.strip()
    .str.lower()
)

customers["customer_state"] = (
    customers["customer_state"]
    .astype(str)
    .str.strip()
    .str.upper()
)

sellers["seller_city"] = (
    sellers["seller_city"]
    .astype(str)
    .str.strip()
    .str.lower()
)

sellers["seller_state"] = (
    sellers["seller_state"]
    .astype(str)
    .str.strip()
    .str.upper()
)

print("Text columns standardized successfully!")
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.days

orders["estimated_delivery_days"] = (
    orders["order_estimated_delivery_date"]
    - orders["order_purchase_timestamp"]
).dt.days

orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.days

print("Delivery features created successfully!")
```

```python
orders["order_year"] = (
    orders["order_purchase_timestamp"]
    .dt.year
)

orders["order_month"] = (
    orders["order_purchase_timestamp"]
    .dt.month
)

orders["order_quarter"] = (
    orders["order_purchase_timestamp"]
    .dt.quarter
)

orders["order_weekday"] = (
    orders["order_purchase_timestamp"]
    .dt.day_name()
)

print("Time-based features created successfully!")
```

```python
orders["delivery_status"] = np.where(
    orders["order_delivered_customer_date"].isna(),
    "Not Delivered",
    np.where(
        orders["order_delivered_customer_date"]
        <= orders["order_estimated_delivery_date"],
        "On Time",
        "Delayed"
    )
)

print("Delivery status created successfully!")
```

```python
products = products.merge(
    category_translation,
    on="product_category_name",
    how="left"
)

products["product_category_name_english"] = (
    products["product_category_name_english"]
    .fillna(products["product_category_name"])
)

print("English product categories added successfully!")
```

```python
print("Missing Values After Cleaning")
print("=" * 80)

for name, dataframe in datasets.items():

    missing_values = dataframe.isnull().sum()

    missing_values = missing_values[missing_values > 0]

    print(f"\n{name}")

    if len(missing_values) == 0:
        print("No missing values found.")

    else:
        print(missing_values)
```

```python
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}

print("Dataset dictionary updated successfully!")
```

```python
customers.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "customers_cleaned.csv"),
    index=False
)

geolocation.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "geolocation_cleaned.csv"),
    index=False
)

order_items.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "order_items_cleaned.csv"),
    index=False
)

order_payments.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "order_payments_cleaned.csv"),
    index=False
)

order_reviews.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "order_reviews_cleaned.csv"),
    index=False
)

orders.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "orders_cleaned.csv"),
    index=False
)

products.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "products_cleaned.csv"),
    index=False
)

sellers.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "sellers_cleaned.csv"),
    index=False
)

category_translation.to_csv(
    os.path.join(PROCESSED_DATA_DIR, "category_translation_cleaned.csv"),
    index=False
)

print("All cleaned datasets saved successfully!")
```
