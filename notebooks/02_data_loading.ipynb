# Notebook 02 — Data Loading

This notebook loads the raw Olist e-commerce datasets into Pandas DataFrames.

## Objectives

- Load all raw CSV files
- Verify that the expected files exist
- Inspect dataset dimensions
- Display sample records
- Check column names and data types
- Check missing values
- Check duplicate records
- Create a dataset summary

The raw data is located in:

`data/raw/`

```python
import os
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
RAW_DATA_DIR = r"C:\Users\asus\Documents\ECOMMERCE-AI-PLATFORM\data\raw"

print("Raw data directory:")
print(RAW_DATA_DIR)
```

```python
if not os.path.isdir(RAW_DATA_DIR):

    raise FileNotFoundError(
        f"Raw data directory not found:\n{RAW_DATA_DIR}"
    )

print("Raw data directory found successfully!")

files = os.listdir(RAW_DATA_DIR)

print(f"Number of items found: {len(files)}")
print("\nFiles available in raw data directory:\n")

for file in files:
    print(file)
```

```python
csv_files = [
    file
    for file in files
    if file.lower().endswith(".csv")
]

print(f"Number of CSV files found: {len(csv_files)}")
```

```python
expected_files = [
    "olist_customers_dataset.csv",
    "olist_geolocation_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_order_payments_dataset.csv",
    "olist_order_reviews_dataset.csv",
    "olist_orders_dataset.csv",
    "olist_products_dataset.csv",
    "olist_sellers_dataset.csv",
    "product_category_name_translation.csv"
]

print(f"Expected number of datasets: {len(expected_files)}")
```

```python
missing_files = [
    file
    for file in expected_files
    if file not in csv_files
]

if missing_files:

    print("The following expected files are missing:\n")

    for file in missing_files:
        print(file)

else:

    print("All expected Olist dataset files are available!")
```

```python
unexpected_files = [
    file
    for file in csv_files
    if file not in expected_files
]

if unexpected_files:

    print("Additional CSV files found:\n")

    for file in unexpected_files:
        print(file)

else:

    print("No additional CSV files found.")
```

```python
customers = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_customers_dataset.csv"
    )
)

print("Customers dataset loaded successfully!")
```

```python
geolocation = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_geolocation_dataset.csv"
    )
)

print("Geolocation dataset loaded successfully!")
```

```python
order_items = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_order_items_dataset.csv"
    )
)

print("Order items dataset loaded successfully!")
```

```python
order_payments = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_order_payments_dataset.csv"
    )
)

print("Order payments dataset loaded successfully!")
```

```python
order_reviews = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_order_reviews_dataset.csv"
    )
)

print("Order reviews dataset loaded successfully!")
```

```python
orders = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_orders_dataset.csv"
    )
)

print("Orders dataset loaded successfully!")
```

```python
products = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_products_dataset.csv"
    )
)

print("Products dataset loaded successfully!")
```

```python
sellers = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "olist_sellers_dataset.csv"
    )
)

print("Sellers dataset loaded successfully!")
```

```python
category_translation = pd.read_csv(
    os.path.join(
        RAW_DATA_DIR,
        "product_category_name_translation.csv"
    )
)

print("Category translation dataset loaded successfully!")
```

```python
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}

print(f"Total datasets loaded: {len(datasets)}")
```

```python
for name, df in datasets.items():

    print(
        f"{name:<25}"
        f"Rows: {df.shape[0]:>8} | "
        f"Columns: {df.shape[1]:>3}"
    )
```

```python
dataset_summary = pd.DataFrame(
    [
        {
            "Dataset": name,
            "Rows": df.shape[0],
            "Columns": df.shape[1]
        }
        for name, df in datasets.items()
    ]
)

dataset_summary
```

```python
for name, df in datasets.items():

    print("=" * 70)
    print(f"DATASET: {name.upper()}")
    print("=" * 70)

    print("Columns:")
    print(list(df.columns))

    print()
```

```python
for name, df in datasets.items():

    print("=" * 70)
    print(f"DATASET: {name.upper()}")
    print("=" * 70)

    display(df.head())

    print()
```

```python
for name, df in datasets.items():

    print("=" * 70)
    print(f"DATASET: {name.upper()}")
    print("=" * 70)

    display(
        df.dtypes.to_frame(
            name="Data Type"
        )
    )

    print()
```

```python
missing_summary = pd.DataFrame(
    [
        {
            "Dataset": name,
            "Missing Values": df.isnull().sum().sum()
        }
        for name, df in datasets.items()
    ]
)

missing_summary
```

```python
duplicate_summary = pd.DataFrame(
    [
        {
            "Dataset": name,
            "Duplicate Rows": df.duplicated().sum()
        }
        for name, df in datasets.items()
    ]
)

duplicate_summary
```

```python
memory_summary = pd.DataFrame(
    [
        {
            "Dataset": name,
            "Memory Usage (MB)": round(
                df.memory_usage(deep=True).sum()
                / (1024 ** 2),
                2
            )
        }
        for name, df in datasets.items()
    ]
)

memory_summary
```

```python
complete_summary = pd.DataFrame(
    [
        {
            "Dataset": name,
            "Rows": df.shape[0],
            "Columns": df.shape[1],
            "Missing Values": df.isnull().sum().sum(),
            "Duplicate Rows": df.duplicated().sum(),
            "Memory Usage (MB)": round(
                df.memory_usage(deep=True).sum()
                / (1024 ** 2),
                2
            )
        }
        for name, df in datasets.items()
    ]
)

complete_summary
```

```python
print("=" * 70)
print("DATA LOADING COMPLETED SUCCESSFULLY")
print("=" * 70)

print(f"\nTotal datasets loaded: {len(datasets)}")

print("\nDatasets loaded:")

for name in datasets:

    print(f"✔ {name}")
```
